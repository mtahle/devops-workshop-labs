# Lab 2 Notebook 3: Validate the Model

Use this notebook to confirm model quality before serving it through the API.

````python
from pathlib import Path
import pickle

import numpy as np
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score

model_path = Path("..") / "model.pkl"
if not model_path.exists():
    raise FileNotFoundError("Run model.ipynb first to generate ../model.pkl")

with model_path.open("rb") as f:
    model = pickle.load(f)

model
````

## Check known samples

````python
cases = [
    ((5.1, 3.5, 1.4, 0.2), "setosa", 0),
    ((6.3, 3.3, 4.7, 1.6), "versicolor", 1),
    ((6.7, 3.0, 5.2, 2.3), "virginica", 2),
]

species = ["setosa", "versicolor", "virginica"]
for features, expected_name, expected_index in cases:
    prediction = int(model.predict(np.array([features]))[0])
    assert prediction == expected_index
    print(features, "->", species[prediction])
````

## Check global accuracy threshold (>= 90%)

````python
X, y = load_iris(return_X_y=True)
accuracy = accuracy_score(y, model.predict(X))
assert accuracy >= 0.90, f"Accuracy {accuracy:.2%} is below the 90% threshold"
print(f"Accuracy check passed: {accuracy:.2%}")
````
